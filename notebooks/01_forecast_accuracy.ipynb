
# Forecast Accuracy Diagnostics

This notebook evaluates forecast accuracy and reliability for PV, wind, load, and other forecast series.


```python
import polars as pl
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from pathlib import Path

# Resolve paths relative to this notebook
base = Path.cwd()
if base.name == "notebooks":
    base = base.parent

cand = [
    base / "data/processed/all_data_transformed.parquet",
    base / "data/processed/all_data_clean.parquet",
    base / "data/processed/all_data.parquet",
]
DATA_PATH = next((p for p in cand if p.exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError("No input parquet found in data/processed/")
```

```python

# Load data
_df = pl.read_parquet(DATA_PATH)
print("rows", _df.height, "cols", len(_df.columns))
```

```python

# Define forecast/actual pairs
pairs = {
    "wind_onshore": ("wind_onshore_forecast", "wind_onshore_actual"),
    "wind_offshore": ("wind_offshore_forecast", "wind_offshore_actual"),
    "solar": ("solar_forecast", "solar_actual"),
    "load": ("load_forecast_da", "load_actual"),
}

# Optional intraday forecast if present
if {"wind_onshore_forecast_intraday", "wind_onshore_actual"}.issubset(_df.columns):
    pairs["wind_onshore_intraday"] = ("wind_onshore_forecast_intraday", "wind_onshore_actual")
```

## Forecast Comparison (Actual vs DA vs ID)

This plot overlays actuals with day-ahead and intraday forecasts for wind and solar.
It prefers ENTSO-E columns when available, otherwise falls back to legacy columns.

```python
# Visualize forecasts vs actuals (prefers ENTSO-E columns)

if "timestamp_utc" not in _df.columns:
    raise KeyError("Missing timestamp_utc")

TECHS = ["wind_onshore", "wind_offshore", "solar"]

# Prefer ENTSO-E columns if present; otherwise fall back to legacy SMARD columns.
def _pick_cols(tech: str):
    entsoe_actual = f"{tech}_actual_entsoe"
    entsoe_da = f"{tech}_forecast_da_entsoe"
    entsoe_id = f"{tech}_forecast_id_entsoe"

    legacy_actual = f"{tech}_actual"
    legacy_da = f"{tech}_forecast"
    legacy_id = f"{tech}_forecast_intraday"

    actual = entsoe_actual if entsoe_actual in _df.columns else legacy_actual
    da = entsoe_da if entsoe_da in _df.columns else legacy_da
    intraday = entsoe_id if entsoe_id in _df.columns else legacy_id

    # Ensure required columns exist
    cols = [c for c in [actual, da, intraday] if c in _df.columns]
    return actual, da, intraday, cols

# Use a recent window for readability (last 30 days)
window_hours = 30 * 24

for tech in TECHS:
    actual, da, intraday, cols = _pick_cols(tech)
    if actual not in _df.columns or da not in _df.columns:
        print(f"[skip] {tech}: missing required columns ({actual}, {da})")
        continue

    plot_cols = ["timestamp_utc"] + cols
    pdf = (
        _df.select(plot_cols)
        .sort("timestamp_utc")
        .tail(window_hours)
        .to_pandas()
    )

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[actual], name=f"{tech} actual", line=dict(width=2)))
    fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[da], name=f"{tech} DA", line=dict(width=1)))
    if intraday in _df.columns:
        fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[intraday], name=f"{tech} ID", line=dict(width=1, dash="dash")))

    fig.update_layout(
        title=f"{tech}: Actual vs DA vs ID (last 30 days)",
        xaxis_title="timestamp_utc",
        yaxis_title="MW",
        legend=dict(orientation="h")
    )
    fig.show()
```

```python

# Compute error metrics

def compute_metrics(df: pl.DataFrame, fc: str, act: str) -> dict:
    d = df.select(["timestamp_utc", fc, act]).drop_nulls()
    if d.height == 0:
        return {"n": 0}
    err = (pl.col(fc) - pl.col(act)).alias("err")
    abs_err = err.abs().alias("abs_err")
    sq_err = (err ** 2).alias("sq_err")
    mape = ((err.abs() / pl.col(act).abs()) * 100).alias("mape")

    out = (
        d.with_columns([err, abs_err, sq_err, mape])
         .select([
            pl.len().alias("n"),
            pl.col("err").mean().alias("ME"),
            pl.col("abs_err").mean().alias("MAE"),
            pl.col("sq_err").mean().alias("MSE"),
            pl.col("sq_err").mean().sqrt().alias("RMSE"),
            pl.col("mape").median().alias("Median_APE"),
        ])
         .to_dicts()[0]
    )
    return out

metrics = []
for name, (fc, act) in pairs.items():
    if fc in _df.columns and act in _df.columns:
        m = compute_metrics(_df, fc, act)
        m["series"] = name
        m["forecast_col"] = fc
        m["actual_col"] = act
        metrics.append(m)

metrics_df = pd.DataFrame(metrics)
metrics_df
```

```python

# Bias by hour-of-day
for name, (fc, act) in pairs.items():
    if fc not in _df.columns or act not in _df.columns:
        continue
    d = (
        _df.select(["timestamp_utc", fc, act])
           .drop_nulls()
           .with_columns((pl.col(fc) - pl.col(act)).alias("err"))
           .with_columns(pl.col("timestamp_utc").dt.hour().alias("hour"))
           .group_by("hour")
           .agg([
               pl.col("err").mean().alias("mean_err"),
               pl.col("err").abs().mean().alias("mae"),
           ])
           .sort("hour")
    )
    fig = px.line(d.to_pandas(), x="hour", y=["mean_err", "mae"], title=f"{name}: Hourly Bias / MAE")
    fig.show()
```

```python

# Seasonal bias by month
for name, (fc, act) in pairs.items():
    if fc not in _df.columns or act not in _df.columns:
        continue
    d = (
        _df.select(["timestamp_utc", fc, act])
           .drop_nulls()
           .with_columns((pl.col(fc) - pl.col(act)).alias("err"))
           .with_columns(pl.col("timestamp_utc").dt.month().alias("month"))
           .group_by("month")
           .agg([
               pl.col("err").mean().alias("mean_err"),
               pl.col("err").abs().mean().alias("mae"),
           ])
           .sort("month")
    )
    fig = px.line(d.to_pandas(), x="month", y=["mean_err", "mae"], title=f"{name}: Monthly Bias / MAE")
    fig.show()
```

```python

# Forecast vs actual scatter (calibration)
for name, (fc, act) in pairs.items():
    if fc not in _df.columns or act not in _df.columns:
        continue
    d = _df.select([fc, act]).drop_nulls().sample(n=min(20000, _df.height), with_replacement=False)
    fig = px.scatter(d.to_pandas(), x=fc, y=act, opacity=0.2, title=f"{name}: Forecast vs Actual")
    fig.add_shape(type="line", x0=d[fc].min(), y0=d[fc].min(), x1=d[fc].max(), y1=d[fc].max(), line=dict(color="red"))
    fig.show()
```

```python

# Rolling error (30-day) to detect instability
window = 30 * 24
for name, (fc, act) in pairs.items():
    if fc not in _df.columns or act not in _df.columns:
        continue
    d = (
        _df.select(["timestamp_utc", fc, act])
           .drop_nulls()
           .with_columns((pl.col(fc) - pl.col(act)).alias("err"))
           .with_columns(pl.col("err").abs().rolling_mean(window).alias("mae_30d"))
           .select(["timestamp_utc", "mae_30d"])
    )
    fig = px.line(d.to_pandas(), x="timestamp_utc", y="mae_30d", title=f"{name}: Rolling 30d MAE")
    fig.show()
```


# Section 4: Market Stress & Uncertainty Feature Engineering


```python

import polars as pl
import plotly.graph_objects as go

# Ensure data is sorted
_df = _df.sort('timestamp_utc')
```

```python

# 1) Feature Engineering (Market Stress & Uncertainty)

# Long-term price volatility (30d rolling std)
if 'da_price_d_eur_mwh' in _df.columns:
    price_col = 'da_price_d_eur_mwh'
elif 'da_price' in _df.columns:
    price_col = 'da_price'
else:
    price_col = None

if price_col:
    _df = _df.with_columns(
        pl.col(price_col).rolling_std(window_size=720, min_samples=1).alias('da_price_rolling_std_30d')
    )

# Short-term error memory (72h rolling MAE)
forecast_pairs = {
    'solar': ('solar_forecast', 'solar_actual'),
    'wind_onshore': ('wind_onshore_forecast', 'wind_onshore_actual'),
    'load': ('load_forecast_da', 'load_actual'),
}
for name, (fc, act) in forecast_pairs.items():
    if fc in _df.columns and act in _df.columns:
        _df = _df.with_columns(
            (pl.col(fc) - pl.col(act)).abs().alias(f'{name}_abs_error')
        )
        _df = _df.with_columns(
            pl.col(f'{name}_abs_error').rolling_mean(window_size=72, min_samples=1).alias(f'{name}_mae_rolling_72h')
        )

# Meteorological ramp intensity (hourly absolute change in forecast)
if 'solar_forecast' in _df.columns:
    _df = _df.with_columns(
        (pl.col('solar_forecast') - pl.col('solar_forecast').shift(1)).abs().alias('solar_ramp')
    )
if 'wind_onshore_forecast' in _df.columns:
    _df = _df.with_columns(
        (pl.col('wind_onshore_forecast') - pl.col('wind_onshore_forecast').shift(1)).abs().alias('wind_ramp')
    )

# Forecast delta: intraday vs day-ahead (wind onshore)
if {'wind_onshore_forecast_intraday', 'wind_onshore_forecast'}.issubset(_df.columns):
    _df = _df.with_columns(
        (pl.col('wind_onshore_forecast_intraday') - pl.col('wind_onshore_forecast')).abs().alias('wind_onshore_delta_fc')
    )

# Residual load forecast
if {'load_forecast_da', 'wind_onshore_forecast', 'solar_forecast'}.issubset(_df.columns):
    _df = _df.with_columns(
        (pl.col('load_forecast_da') - pl.col('wind_onshore_forecast') - pl.col('solar_forecast')).alias('residual_load_forecast')
    )

# Fill rolling nulls for ML usability
fill_cols = [c for c in [
    'da_price_rolling_std_30d',
    'solar_mae_rolling_72h', 'wind_onshore_mae_rolling_72h', 'load_mae_rolling_72h',
    'solar_ramp', 'wind_ramp', 'wind_onshore_delta_fc', 'residual_load_forecast'
] if c in _df.columns]
if fill_cols:
    _df = _df.with_columns([pl.col(c).forward_fill().backward_fill().alias(c) for c in fill_cols])
```

```python

# 2) Crisis Indicator Logic

if 'da_price_rolling_std_30d' in _df.columns:
    price_thr = _df.select(pl.col('da_price_rolling_std_30d').quantile(0.9)).item()
else:
    price_thr = None

# pick solar_mae if present, else wind_mae
unc_col = None
for c in ['solar_mae_rolling_72h', 'wind_onshore_mae_rolling_72h']:
    if c in _df.columns:
        unc_col = c
        break

if price_thr is not None and unc_col is not None:
    unc_thr = _df.select(pl.col(unc_col).quantile(0.9)).item()
    _df = _df.with_columns(
        ((pl.col('da_price_rolling_std_30d') >= price_thr) & (pl.col(unc_col) >= unc_thr)).alias('is_high_uncertainty')
    )
```

```python

# 3) Visual analytics: rolling volatility vs rolling MAE + crisis shading

plot_cols = [c for c in ['da_price_rolling_std_30d', 'solar_mae_rolling_72h', 'wind_onshore_mae_rolling_72h', 'load_mae_rolling_72h'] if c in _df.columns]
plot_df = _df.select(['timestamp_utc'] + plot_cols)

fig = go.Figure()
if 'da_price_rolling_std_30d' in plot_df.columns:
    fig.add_trace(go.Scatter(x=plot_df['timestamp_utc'], y=plot_df['da_price_rolling_std_30d'], name='Price Volatility (30d)'))

for c in plot_cols:
    if c != 'da_price_rolling_std_30d':
        fig.add_trace(go.Scatter(x=plot_df['timestamp_utc'], y=plot_df[c], name=c))

# Crisis shading
if 'is_high_uncertainty' in _df.columns:
    crisis = _df.select(['timestamp_utc', 'is_high_uncertainty']).to_pandas()
    in_block = False
    start = None
    for ts, flag in zip(crisis['timestamp_utc'], crisis['is_high_uncertainty']):
        if flag and not in_block:
            start = ts
            in_block = True
        elif not flag and in_block:
            fig.add_vrect(x0=start, x1=ts, fillcolor='red', opacity=0.15, line_width=0)
            in_block = False

fig.update_layout(title='Market Stress & Forecast Uncertainty', xaxis_title='Time')
fig.show()
```

## Summary of Findings

**1. Data Patterns & Seasonality**

- **Solar Seasonality (Spring Peak in Errors):**
- Peak Error in April: Solar reliability drops significantly in Spring. MAE peaks in April (723 MW) and March (598 MW), compared to winter lows (~230 MW).
- Negative Bias: A strong negative bias exists from March–May (underpredicting by up to 380 MW). The forecast consistently underestimates the "Spring surge," leading to potential unexpected negative price spikes.

- **Winter Volatility (Wind & Load):**
- Wind Onshore: MAE peaks in January (1,479 MW) and December (1,300 MW), driven by winter storms which are difficult to forecast in timing and intensity.
- Load: Forecast errors also peak in January due to high heating demand and holiday-related behavior shifts.

- **Diurnal (Hourly) Patterns:**
- Solar "Noon Peak": Errors scale with sun intensity, creating the most "noise" between 10:00 and 14:00.
- Load "Ramp-up" Error: Bias increases during morning (06:00–09:00) and evening (17:00–20:00) ramps, where forecasts struggle to match the speed of demand changes.

- **Implications:**
- The "Solar Weak Link": High solar error (41%) correlates with negative prices.
- Seasonal Features: The ML model must include "Month" or "Season" to adjust for these shifting accuracy profiles.

**2. Data Usability & Reliability (APE Metrics)**

- **Load (High Reliability):** Median APE ~3.19%. "Gold standard" feature; stable baseline for behavior prediction.
- **Wind Onshore (Reliable):** Median APE ~9.12%. Robust enough to serve as a primary feature.
- **Wind Offshore (Moderate Reliability):** Median APE ~16.52%. Usable, but ML models must account for higher uncertainty during storm events.
- **Solar (Low Reliability):** Median APE ~41.17%. Very poor accuracy. Using this as a raw feature may introduce significant noise without feature engineering.
- **Wind Onshore Intraday (Critical Issue - Pre-Fix):**
- Note: Initial analysis showed a Median APE of 215% and Mean Error of ~18,000 MW. This indicated a unit scaling bug (Sum vs Mean) which has since been addressed in the pipeline.


## SMARD vs ENTSO-E Forecast Accuracy (Wind & Solar)

This section compares SMARD vs ENTSO-E day-ahead forecasts against actuals, and then overlays both forecasts together.

```python
import pandas as pd
import plotly.graph_objects as go

if "timestamp_utc" not in _df.columns:
    raise KeyError("Missing timestamp_utc")

TECHS = ["wind_onshore", "wind_offshore", "solar"]

# Column mapping
COLS = {
    "smard": {
        "actual": {
            "wind_onshore": "wind_onshore_actual",
            "wind_offshore": "wind_offshore_actual",
            "solar": "solar_actual",
        },
        "forecast_da": {
            "wind_onshore": "wind_onshore_forecast",
            "wind_offshore": "wind_offshore_forecast",
            "solar": "solar_forecast",
        },
        "forecast_id": {
            "wind_onshore": "wind_onshore_forecast_intraday",
            "wind_offshore": "wind_offshore_forecast_intraday",
            "solar": "solar_forecast_intraday",
        },
    },
    "entsoe": {
        "actual": {
            "wind_onshore": "wind_onshore_actual_entsoe",
            "wind_offshore": "wind_offshore_actual_entsoe",
            "solar": "solar_actual_entsoe",
        },
        "forecast_da": {
            "wind_onshore": "wind_onshore_forecast_da_entsoe",
            "wind_offshore": "wind_offshore_forecast_da_entsoe",
            "solar": "solar_forecast_da_entsoe",
        },
        "forecast_id": {
            "wind_onshore": "wind_onshore_forecast_id_entsoe",
            "wind_offshore": "wind_offshore_forecast_id_entsoe",
            "solar": "solar_forecast_id_entsoe",
        },
    },
}

# Use last 30 days for plotting readability
window_hours = 30 * 24

for tech in TECHS:
    sm_act = COLS["smard"]["actual"][tech]
    sm_da = COLS["smard"]["forecast_da"][tech]
    sm_id = COLS["smard"]["forecast_id"][tech]
    en_act = COLS["entsoe"]["actual"][tech]
    en_da = COLS["entsoe"]["forecast_da"][tech]
    en_id = COLS["entsoe"]["forecast_id"][tech]

    # Prefer ENTSO-E actual if available, else SMARD actual
    actual_col = en_act if en_act in _df.columns else sm_act

    needed = ["timestamp_utc", actual_col]
    for c in [sm_da, sm_id, en_da, en_id]:
        if c in _df.columns:
            needed.append(c)

    if len(needed) <= 2:
        print(f"[skip] {tech}: missing forecasts")
        continue

    pdf = (
        _df.select(needed)
        .sort("timestamp_utc")
        .tail(window_hours)
        .to_pandas()
    )

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[actual_col], name="Actual", line=dict(width=2)))
    if sm_da in pdf.columns:
        fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[sm_da], name="SMARD DA", line=dict(width=1)))
    if sm_id in pdf.columns:
        fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[sm_id], name="SMARD ID", line=dict(width=1, dash="dot")))
    if en_da in pdf.columns:
        fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[en_da], name="ENTSO-E DA", line=dict(width=1, dash="dash")))
    if en_id in pdf.columns:
        fig.add_trace(go.Scatter(x=pdf["timestamp_utc"], y=pdf[en_id], name="ENTSO-E ID", line=dict(width=1, dash="dashdot")))

    fig.update_layout(
        title=f"{tech} — SMARD vs ENTSO-E (DA/ID vs Actual, last 30 days)",
        xaxis_title="timestamp_utc",
        yaxis_title="MW",
        legend=dict(orientation="h")
    )
    fig.show()

# Error comparison (MAE)
rows = []
for tech in TECHS:
    sm_act = COLS["smard"]["actual"][tech]
    sm_da = COLS["smard"]["forecast_da"][tech]
    sm_id = COLS["smard"]["forecast_id"][tech]
    en_act = COLS["entsoe"]["actual"][tech]
    en_da = COLS["entsoe"]["forecast_da"][tech]
    en_id = COLS["entsoe"]["forecast_id"][tech]

    if sm_act in _df.columns and sm_da in _df.columns:
        mae = _df.select((pl.col(sm_da) - pl.col(sm_act)).abs().mean().alias("mae"))["mae"][0]
        rows.append({"tech": tech, "source": "smard_da", "mae": float(mae)})
    if sm_act in _df.columns and sm_id in _df.columns:
        mae = _df.select((pl.col(sm_id) - pl.col(sm_act)).abs().mean().alias("mae"))["mae"][0]
        rows.append({"tech": tech, "source": "smard_id", "mae": float(mae)})

    if en_act in _df.columns and en_da in _df.columns:
        mae = _df.select((pl.col(en_da) - pl.col(en_act)).abs().mean().alias("mae"))["mae"][0]
        rows.append({"tech": tech, "source": "entsoe_da", "mae": float(mae)})
    if en_act in _df.columns and en_id in _df.columns:
        mae = _df.select((pl.col(en_id) - pl.col(en_act)).abs().mean().alias("mae"))["mae"][0]
        rows.append({"tech": tech, "source": "entsoe_id", "mae": float(mae)})

err_df = pd.DataFrame(rows)
print(err_df)

if not err_df.empty:
    fig = go.Figure()
    for tech in TECHS:
        sub = err_df[err_df["tech"] == tech]
        if sub.empty:
            continue
        fig.add_trace(go.Bar(x=sub["source"], y=sub["mae"], name=tech))
    fig.update_layout(title="MAE Comparison: SMARD vs ENTSO-E (DA/ID)", barmode="group", xaxis_title="source", yaxis_title="MAE (MW)")
    fig.show()
```
