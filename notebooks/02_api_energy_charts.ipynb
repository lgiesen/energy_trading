# Energy-Charts

[Documentation](https://api.energy-charts.info/)

Required data:
- Day-Ahead electricity price (DE/LU) $P^{DA}_{t}$
- Day-Ahead electricity prices of neighboring bidding zones $P^{DA,\text{NB}}_{t}$
(AT, BE, CH, CZ, DK1, DK2, FR, NL, NO2, PL, SE4)

Script at `energy_trading/ingestion/fetch_energy_charts_prices.py` and output of the script at `data/day_ahead_prices.parquet`. Run the script with:

```python -m energy_trading.ingestion.fetch_energy_charts_prices --start 2022-01-01T00:00Z --end 2022-01-01T23:00Z --out data/day_ahead_prices.parquet --timeout 120```

````python
import requests

start = "2022-01-01T00:00Z"
end = "2022-01-02T00:00Z"
base_url = f"https://api.energy-charts.info/price?start={start}&end={end}"

payload = {}
files={}
headers = {}

region = "DE-LU"
url = f"{base_url}&bzn={region}"
response = requests.request("GET", url, headers=headers, data=payload, files=files)

print(response.text)
````
