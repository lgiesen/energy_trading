```python
import io
import pandas as pd


def fetch_table(url: str) -> pd.DataFrame:
    resp = requests.get(url, headers={"Authorization": f"{bearer_token}"}, timeout=60)
    resp.raise_for_status()
    return pd.read_csv(io.StringIO(resp.text), sep=";")


def tidy_nrv(df: pd.DataFrame) -> pd.DataFrame:
    val_cols = [c for c in df.columns if "deutsch" in c.lower() or "saldo" in c.lower()]
    value_col = val_cols[0] if val_cols else df.columns[-1]
    ts = pd.to_datetime(df["Datum"] + " " + df["von"], format="%d.%m.%Y %H:%M", utc=True)
    vals = pd.to_numeric(
        df[value_col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .replace({"N.A.": pd.NA}),
        errors="coerce",
    )
    out = pd.DataFrame(
        {
            "timestamp_utc": ts,
            "NRV_balance": vals,
            "Datum": df.get("Datum"),
            "Zeitzone": df.get("Zeitzone"),
            "time": df.get("von"),
        }
    )
    return out


def tidy_rebap(df: pd.DataFrame) -> pd.DataFrame:
    unter_cols = [c for c in df.columns if "unter" in c.lower()]
    ueber_cols = [c for c in df.columns if "ueber" in c.lower()]
    unter_col = unter_cols[0] if unter_cols else df.columns[-2]
    ueber_col = ueber_cols[0] if ueber_cols else df.columns[-1]
    ts = pd.to_datetime(df["Datum"] + " " + df["von"], format="%d.%m.%Y %H:%M", utc=True)
    unter = pd.to_numeric(
        df[unter_col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .replace({"N.A.": pd.NA}),
        errors="coerce",
    )
    ueber = pd.to_numeric(
        df[ueber_col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .replace({"N.A.": pd.NA}),
        errors="coerce",
    )
    out = pd.DataFrame(
        {
            "timestamp_utc": ts,
            "reBAP unterdeckt": unter,
            "reBAP ueberdeckt": ueber,
            "Datum": df.get("Datum"),
            "Zeitzone": df.get("Zeitzone"),
            "time": df.get("von"),
        }
    )
    return out


nrv_url = f"{base_url}/NrvSaldo/NRVSaldo/Qualitaetsgesichert/{start}/{end}"
rebap_url = f"{base_url}/NrvSaldo/reBAP/Qualitaetsgesichert/{start}/{end}"

nrv_raw = fetch_table(nrv_url)
rebap_raw = fetch_table(rebap_url)

nrv_df = tidy_nrv(nrv_raw)
rebap_df = tidy_rebap(rebap_raw)

merged = pd.merge(nrv_df, rebap_df, on="timestamp_utc", how="outer", suffixes=("", "_rebap"))

# Drop duplicate/meta columns and TSO-specific columns not needed for ML
meta_drop = [c for c in merged.columns if c in ("Datum", "Zeitzone", "time", "Datum_rebap", "Zeitzone_rebap", "time_rebap")]
tso_drop = [
    "50Hertz (Positiv)",
    "Amprion (Positiv)",
    "TenneT TSO (Positiv)",
    "TransnetBW (Positiv)",
    "50Hertz (Negativ)",
    "Amprion (Negativ)",
    "TenneT TSO (Negativ)",
    "TransnetBW (Negativ)",
    "50Hertz",
    "Amprion",
    "TenneT TSO",
    "TransnetBW",
]
merged = merged.drop(columns=[c for c in meta_drop + tso_drop if c in merged.columns])

merged = merged.sort_values("timestamp_utc")
merged.head()
```

# Activation Rate

```python
import io
import pandas as pd
from datetime import timedelta

def _parse_activation_csv(raw_text: str, prefix: str) -> pd.DataFrame:
    cleaned = raw_text.lstrip('\ufeff').lstrip('ï»¿').replace('\r\n', '\n').replace('\r', '\n')
    df = pd.read_csv(io.StringIO(cleaned), sep=';', decimal=',', low_memory=False)
    df.columns = df.columns.str.strip().str.lower()
    date_col = 'datum' if 'datum' in df.columns else df.columns[0]
    time_col = 'von' if 'von' in df.columns else df.columns[1]
    ts = pd.to_datetime(df[date_col] + ' ' + df[time_col], format='%d.%m.%Y %H:%M', errors='coerce')
    ts = ts.dt.tz_localize('UTC')
    pos_col = next((c for c in df.columns if 'positiv' in c), None)
    neg_col = next((c for c in df.columns if 'negativ' in c), None)
    if not pos_col or not neg_col:
        raise ValueError('Positive/negative columns not found in activation CSV')
    def clean_num(s):
        return (
            s.astype(str)
             .str.replace('.', '', regex=False)
             .str.replace(',', '.', regex=False)
        )
    out = pd.DataFrame({
        'timestamp_utc': ts,
        f'{prefix}_activated_mw_pos': pd.to_numeric(clean_num(df[pos_col]), errors='coerce'),
        f'{prefix}_activated_mw_neg': pd.to_numeric(clean_num(df[neg_col]), errors='coerce'),
    })
    return out.dropna(subset=['timestamp_utc']).set_index('timestamp_utc').sort_index()

def _fetch_activation(url, bearer_token):
    headers = {'Authorization': f'{bearer_token}', 'Accept': 'text/csv'}
    resp = requests.get(url, headers=headers)
    resp.raise_for_status()
    return resp.text

def get_activated_series(kind: str, start_date: str, end_date: str, bearer_token: str) -> pd.DataFrame:
    endpoint = 'AktivierteSRL' if kind == 'afrr' else 'AktivierteMRL'
    url = f"{base_url}/NrvSaldo/{endpoint}/Qualitaetsgesichert/{start_date}/{end_date}"
    text = _fetch_activation(url, bearer_token)
    return _parse_activation_csv(text, kind)

def get_activated_series_chunked(kind: str, start_ts: pd.Timestamp, end_ts: pd.Timestamp, bearer_token: str, days: int = 180):
    cur = start_ts
    frames = []
    while cur < end_ts:
        nxt = min(cur + timedelta(days=days), end_ts)
        s = cur.isoformat(timespec='minutes')
        e = nxt.isoformat(timespec='minutes')
        try:
            frames.append(get_activated_series(kind, s, e, bearer_token))
        except Exception as exc:
            print(f'Warning: {kind} chunk {s}->{e} failed: {exc}')
        cur = nxt
    if not frames:
        return pd.DataFrame()
    return pd.concat(frames).sort_index().drop_duplicates()

start_ts = pd.to_datetime(start).tz_localize('UTC')
end_ts = pd.to_datetime(end).tz_localize('UTC')

afrr_activated = get_activated_series_chunked('afrr', start_ts, end_ts, bearer_token)
mfrr_activated = get_activated_series_chunked('mfrr', start_ts, end_ts, bearer_token)

activation_df = afrr_activated.join(mfrr_activated, how='outer')
activation_df.head()
```
